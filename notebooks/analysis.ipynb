# Proyecto en dataset de ventas
Este proyecto tiene la misión de analizar las ventas del dataset descargado a partir de https://chatgpt.com/c/69c6a0d8-a034-83e9-a2f3-363149b23731.
El analisis está enfocado en el descubrimiento de tendencias, patrones que puedan ser de beneficio para la toma de decisiones de la superstore

### Importaciones

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
```

### Carga de los datos

```python
# Sube 1 nivel desde el notebook y luego va a lo siguiente de la ruta
ruta = r'..\data\raw\train.csv'
df_inicial = pd.read_csv(ruta)
```

### Ver información inicial de los datos

```python
print(f'\n\nDataFrame inicial:\n{df_inicial}')
print('\n\nData info:')
df_inicial.info()
print(f'\n\nDescribe del DataFrame:\n{df_inicial.describe()}')
```

### Limpieza de datos y tansformaciones
Hay datos faltantes en Postal Code, asi que esos registros seran eliminados, son pocos asi que no hay inconveniente.            
Además las fechas no son un datetime, asi que eso tambien se va a cambiar

```python
df_sin_nulos = df_inicial.dropna(subset=['Postal Code']).copy() # Crear una copia del dataframe original
df_sin_nulos['Order Date'] = pd.to_datetime(df_sin_nulos['Order Date'], format='%d/%m/%Y')
```

### Guardar la información en processed

```python
df_sin_nulos.to_csv('../data/processed/train_clean.csv')
```

### Crear datos necesarios para los gráficos
En primer lugar lo que se hace es un DataFrame para poder manejar la cantidad de veces que se repite una sub-categoria, pudiendo saber cual es la categoria que mas se compra

```python
df_sub_category_count = df_sin_nulos.groupby('Sub-Category').size()
df_sub_category_count = df_sub_category_count.reset_index()
df_sub_category_count = df_sub_category_count.rename(columns={0: 'Count'})
print(f'\n\nSub-category count:\n{df_sub_category_count}')
```

Ahora otro DataFrame para poder controlar cuanto ha vendido en total cada región

```python
df_region_total_sale = df_sin_nulos.groupby('Region')['Sales'].sum()
df_region_total_sale = df_region_total_sale.reset_index()
df_region_total_sale = df_region_total_sale.rename(columns={0: 'Total sale of region'})
df_region_total_sale = df_region_total_sale.sort_values('Sales', ascending=False)
print(f'\n\nRegiones con su venta total:\n{df_region_total_sale}')
```

DataFrame para ver las ventas totales de cada mes

```python
df_months_and_total_sale = df_sin_nulos.groupby(df_sin_nulos['Order Date'].dt.to_period('M'))['Sales'].sum()
df_months_and_total_sale = df_months_and_total_sale.reset_index()
df_months_and_total_sale['Order Date'] = df_months_and_total_sale['Order Date'].astype(str)
df_months_and_total_sale['Order Date'] = pd.to_datetime(df_months_and_total_sale['Order Date'], format='%Y-%m')
print(f'\n\nFecha por mes y su venta total')
```
